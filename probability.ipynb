```python
import pandas as pd

data = {
    "Student" : ["A","B","C","D","E","F","G","H","I","J"],
    "Passed" : ["Yes","No","Yes","Yes","No",
                "Yes","Yes","No","Yes","Yes"]
}
df = pd.DataFrame(data)
df
```

```python
total = len(df)
total
```

```python
passed = len(df[df['Passed']=='Yes'])
```

```python
probability = passed/total
print(probability)
```

```python
failed = len(df[df['Passed']=='No'])
```

```python
prob_failed = failed/total
print(prob_failed)
```

```python
data = {
    "Student" : ["A","B","C","D","E","F"],
    "Gender" : ["Boy","Boy","Boy","Girl","Girl","Girl"],
    "Passed" : ["Yes","No","Yes","Yes","No","Yes"]
}
df = pd.DataFrame(data)
df
```

### Probability of boys or passed

```python
total= len(df)
```

```python
len(df[(df['Gender']=='Boy') | (df['Passed']=='Yes')])/total
```

### Probability of girls or passed

```python
len(df[(df['Gender']=='Girl') | (df['Passed']=='Yes')])/total
```

### Probability of girls and passed

```python
len(df[(df['Gender']=='Girl') & (df['Passed']=='No')])/total
```

### Probability of boy or passed

```python
len(df[(df['Gender']=='Boy') & (df['Passed']=='Yes')])/total
```
