```python
import pandas as pd

data = {
    "Student" : ["A","B","C","D","E","F","G","H","I","J"],
    "Cricket" : ["Yes","Yes","Yes","No","No","Yes","No","Yes","No","Yes"],
    "Passed" : ["Yes","Yes","No","Yes","No","Yes","No","Yes","Yes","No"]
}
df = pd.DataFrame(data)
df
```

### Students who played cricket?

```python
total = len(df)
```

```python
play_cricket = len(df[df['Cricket']=='Yes'])/total
play_cricket
```

### Students who passed?

```python
passed = len(df[df['Passed']=='Yes'])/total
passed
```

### Cricket and passed

```python
cricket_passed = len(df[(df['Cricket']=='Yes') & (df['Passed']=='Yes')])/total
cricket_passed
```

### Cricket or passed

```python
cricket_or_passed = len(df[(df['Cricket']=='Yes') | (df['Passed']=='Yes')])/total
cricket_or_passed
```

### Probability of passed given he plays cricket

```python
passed_given_cricket = cricket_passed/play_cricket                 # P(P/C) = P(PnC)/P(C)
```

```python
passed_given_cricket
```

### Bayes Theorem - probability of cricket given passed

```python
cricket_given_passed = passed_given_cricket*play_cricket/passed                   # P(C/P) = P(P/C).P(C)/P(P)
cricket_given_passed
```

```python
data = {
    "Student" : ["A","B","C","D","E"],
    "Study_hours" : [2,4,6,8,10]
}
df = pd.DataFrame(data)
df
```

```python
# we dont have a direct formula to calculate mean absolute deviation sowe do it step by step wiyh formula.
# xi-meu/n
```

```python
mean_value = df['Study_hours'].mean()            # calculted mean (meu)
mean_value
```

```python
df['Deviation'] = df['Study_hours']- mean_value            # xi-meu
```

```python
df
```

```python
df['Absolute_Deviation'] = abs(df['Deviation'])           # calculated absolute deviation
df
```

```python
mean_absolute_deviation = df['Absolute_Deviation'].mean()
mean_absolute_deviation
```
